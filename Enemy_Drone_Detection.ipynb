```python
# image = np.zeros((5, 5, 3))
#                  │  │  │
#                  │  │  └── 3 color channels
#                  │  └───── width
#                  └──────── height


# Think of dtype=np.uint8 as telling NumPy what kind of numbers we want to store.

# dtype=np.uint8

# Break it down:

# dtype → short for data type. It tells NumPy what type of data to store.
# np → NumPy.
# uint8 → a type of integer number.

# uint8 means:

# u = unsigned → no negative numbers
# int = integer → whole numbers, not decimals
# 8 = uses 8 bits to store the number
# Therefore, the range is 0 to 255

# So:

# dtype=np.uint8

# basically means:

# "Store these values as whole numbers between 0 and 255."

# For images, this is useful because pixel color values are commonly represented in this range:

# 0   → minimum intensity
# 255 → maximum intensity

# You don't need to understand the 8-bit concept deeply yet. For our computer-vision project, just remember:

# uint8 = non-negative whole numbers from 0–255.
```

```python
# WHAT WE ARE DOING NOW: Cropping a specific region of the drone image using slicing.
import cv2

# Read the image file.
image = cv2.imread("test_image.jpg")

# Remember: image[Y, X] is the same as image[Row, Column]
# Let's select a box around the drone. 
# These numbers are just guesses to cut out the middle part.
y_start = 200  # Start at row 200
y_end = 800    # Stop before row 800
x_start = 600  # Start at column 600
x_end = 1300   # Stop before column 1300

# Slice the image to get only that box.
# This creates a new, smaller image.
cropped_drone = image[y_start:y_end, x_start:x_end]

# Display the cropped image in a new window.
cv2.imshow("Cropped Drone", cropped_drone)

# Wait for a key press.
cv2.waitKey(0)

# Close the window.
cv2.destroyAllWindows()
```

```python
# WHAT WE ARE DOING NOW: Understanding how we guess the bounding box coordinates manually.
import cv2

# Read the original image. 
# We know its shape is (1080 rows, 1920 columns).
image = cv2.imread("test_image.jpg")

# The image starts at row 0 (top) and column 0 (left).
# The image ends at row 1079 (bottom) and column 1919 (right).

# We haven't trained YOLO yet, so we are doing this manually to understand the logic.
# These numbers are just our best guesses by looking at the picture.

# 1. Top edge of the drone: roughly 200 pixels down from the top.
row_top = 200        

# 2. Left edge of the drone: roughly 600 pixels from the left side.
column_left = 600    

# 3. Bottom edge of the drone: roughly 800 pixels down.
row_bottom = 800     

# 4. Right edge of the drone: roughly 1300 pixels from the left side.
column_right = 1300  

# Draw the rectangle using these guessed numbers.
# (column_left, row_top) is the top-left corner.
# (column_right, row_bottom) is the bottom-right corner.
# 5 is the thickness of the line.
cv2.rectangle(image, (column_left, row_top), (column_right, row_bottom), (0, 255, 0), 5)

# Show the image with the box.
cv2.imshow("Drone with Bounding Box", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# WHAT WE ARE DOING NOW: Reading live video frames from your webcam.
import cv2

# Open the default camera (0 usually means the built-in webcam).
cap = cv2.VideoCapture(0)

# Check if the camera opened successfully.
if not cap.isOpened():
    print("Cannot open camera")

# Start an infinite loop to read the video frame by frame.
while True:
    # Read one frame. 'ret' is True if the frame was read successfully.
    # 'frame' is the actual picture (a NumPy array).
    ret, frame = cap.read()
    
    # If reading failed (e.g., camera disconnected), break the loop.
    if not ret:
        print("Can't receive frame. Exiting...")
        break

    # Display the current frame in a window.
    cv2.imshow("My Drone Camera Feed", frame)

    # We haven't trained YOLO yet, so we are doing this manually to understand the logic.
    # Wait 1 millisecond for a key press. If the 'q' key is pressed, break the loop to stop the video.
    if cv2.waitKey(1) == ord('q'):
        break

# Release the camera and close all windows when done.
cap.release()
cv2.destroyAllWindows()
```

```python
# WHAT WE ARE DOING NOW: Drawing a box and alert text on a moving video frame.
import cv2

# Open the video file.
cap = cv2.VideoCapture("test_video.mp4")

# Check if the video opened successfully.
if not cap.isOpened():
    print("Cannot open video file")

# Start the video loop.
while True:
    # Read one frame from the video.
    ret, frame = cap.read()
    
    # If the video ended, break the loop.
    if not ret:
        print("Video finished. Exiting...")
        break

    # --- Draw a green box in the center (Targeting Reticle) ---
    box_top_left = (700, 400)
    box_bottom_right = (1200, 900)
    cv2.rectangle(frame, box_top_left, box_bottom_right, (0, 255, 0), 3)    #BGR

    # --- Add a text alert above the box ---
    text_position = (700, 380) # Just above the box
    
    # cv2.putText(image, text, position, font, font_scale, color, thickness)
    # font: cv2.FONT_HERSHEY_SIMPLEX is the font style (like Arial).
    # font_scale: 1.0 is normal size. 2.0 is double size.
    # color: (0, 0, 255) is Red in BGR format (0 Blue, 0 Green, 255 Red).
    # thickness: 2 is how thick the text is (like using a thicker marker).
    cv2.putText(frame, "ALERT: SCANNING FOR DRONES...", text_position, 
                cv2.FONT_HERSHEY_SIMPLEX, 1.0, (0, 0, 255), 2)

    # Display the frame with the box and text.
    cv2.imshow("Enemy Drone Detection System", frame)

    # Wait 25 milliseconds, press 'q' to quit.
    if cv2.waitKey(25) == ord('q'):
        break

# Release and close.
cap.release()
cv2.destroyAllWindows()
```

```python
# What does YOLO actually give you?
# When you feed YOLO a picture (or a video frame), it gives you three things:

# A Bounding Box: It gives you the exact numbers for column_left, row_top, column_right, and row_bottom of the object it found. (This is the exact same data we were guessing manually!)

# A Class: It tells you what it thinks the object is. (e.g., "Drone", "Bird", "Person").

# A Confidence Score: It tells you how sure it is. (e.g., "I am 95% sure this is a drone").
```

```python
# WHAT WE ARE DOING NOW: Using YOLO to automatically detect objects in our drone image.
from ultralytics import YOLO

# 1. Load the pre-trained YOLO model.
# 'yolov8n.pt' is the 'nano' version (smallest and fastest, best for beginners).
model = YOLO("yolov8n.pt")

# 2. Ask YOLO to look at your drone image.
# YOLO returns a list of results, because it can process multiple images at once.
results = model("test_image.jpg")

# 3. Access the first (and only) result from the list.
# In Python, indexing starts at 0. So results[0] means: "Give me the result for the first image."
# This result contains the bounding boxes, class names, and confidence scores.
# .show() then displays the image with YOLO's automatic boxes drawn on it.
results[0].show()
```

```python
# WHAT WE ARE DOING NOW: Running YOLO on every frame of our drone video.
import cv2
from ultralytics import YOLO

# 1. Load the pre-trained YOLO model.
model = YOLO("yolov8n.pt")

# 2. Open the video file.
cap = cv2.VideoCapture("test_video.mp4")

# 3. Start the video loop.
while True:
    # Read one frame from the video.
    ret, frame = cap.read()
    
    # If the video ended, break the loop.
    if not ret:
        print("Video finished. Exiting...")
        break

    # 4. Pass the current frame into YOLO.
    # YOLO returns a list of results. results[0] is the result for this specific frame.
    results = model(frame)

    # 5. YOLO has a built-in function called .plot()
    # It takes the original frame, draws all the boxes and labels, and returns a new image.
    annotated_frame = results[0].plot()

    # 6. Display the annotated frame.
    cv2.imshow("Enemy Drone Detection System", annotated_frame)

    # Wait 1 millisecond, press 'q' to quit.
    if cv2.waitKey(1) == ord('q'):
        break

# Release and close.
cap.release()
cv2.destroyAllWindows()
```
